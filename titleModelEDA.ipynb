```python
import tarfile
import pandas as pd
from pathlib import Path

downloads = Path.home() / "Downloads"
extract_dir = downloads / "booksummaries_extracted"

if not extract_dir.exists():
    with tarfile.open(downloads / "booksummaries.tar.gz") as tar:
        tar.extractall(extract_dir)

columns = ["wikipedia_id", "freebase_id", "title", "author", "pub_date", "genres", "summary"]
df = pd.read_csv(
    extract_dir / "booksummaries" / "booksummaries.txt",
    sep="\t",
    names=columns,
)

df.head()
```
